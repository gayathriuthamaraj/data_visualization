# **API**

Application Programming Interface

Set of communication rules and  abilities

Enables interactions between software applications

# **Web APIs, clients and servers**

Web APIs communicate over the internet using HTTP protocol

## **Types of Web APIs**

### **1. SOAP (Simple Object Access Protocol)**

Focus on strict and formal API design

Enterprise applications(bank, government, insurance, legacy)

Secure, transactional, or legacy system

Has built-in error handling, security standards (WS-Security), and strict contracts

XML

### **2. REST (Representational State Transfer)**

Simple CRUD operations (Create, Read, Update, Delete)

Focus on simplicity and scalability

Most common API architecture

Wide support and easy to cache

Usually JSON(sometimes XML, CSV, etc.)

### **3. GraphQL**

Avoids over-fetching and under-fetching

Focus on flexibility

Optimized for performance

Great for complex frontends

Single endpoint handles all queries

JSON

## **urllib**

Bundled with Python

Powerful but not very developer-friendly

Good for system scripts and low-level control

To use when, you need fine-grained control over request behavior

```
from urllib.request import urlopen
api = "http://api.music-catalog.com/"

    data = response.read()

```

## **requests**

Many powerful built-in features

Easier to use

Automatic data handling 

You're working with web APIs (REST, GraphQL, etc.)

You'll handle authentication or sessions

You want to send headers, query params or JSON easily


## **URL**

URL = Uniform Resource Locator

The structured address to an API Resource

Customize the URL to interact with specific API Resources

```
http://350.5th-ave.com/unit/243
```

```
http://
``` 
- Protocol = the means of transportation

```
350.5th-ave.com
``` 
- Domain = the street address of the office building

```
:80
``` 
- Port = the gate or door to use when entering the building

```
/unit/243
``` 
- Path = the specific office unit inside the building

```
?floor = 77
``` 
- Query = any additional instructions

### **Adding query parameters with requests**

```
response = requests.get("http://350.5th-ave.com/unit/243?floor=77&elevator=True")
print(response.url)
```

Using the **params** argument to add query parameters

```
query_params = {"floor":77, "elevator":True}
response = requests.get("http://350.5th-ave.com/unit/243", params = query_params)
print(response.url)
```

**GET** - Read - Check the mailbox contents

**POST** - Create - Drop a new package in the mailbox

**PUT** - Update - Replace all packages with a new one

**DELETE** - Delete - Remove all packages from the mailbox

### **GET** 
- Retrieve data from server
- Body: No body (parameters go in URL)

```
GET /users/101
```

### **POST**
- Send new data to the server
- Body: Usually contains data (JSON, form data)

```
POST /users
Content-Type: application/json

{
  "name": "Gayathri",
  "role": "Your friend next-door"
}
```

### **PUT**
- Update or replace an existing resource
- Body: Required - Full object(not partial)

```
PUT /users/101
{
  "name": "Gayathri Uthamaraj",
  "role": "Hello"
}
```
### **PATCH**
- Update part of an existing resource

```
PATCH /users/101
{
  "role": "Manager"
}
```

### **DELETE**
- Remove a resource completely

```
DELETE /users/101
```

### **HEAD**
- Same as GET but returns only headers
- Checking if a resource exists, or metadata

```
HEAD /file.pdf
```

### **OPTIONS**
- Asks the server which methods are allowed
- Prefight checks in CORS or debugging API

```
OPTIONS /users
```

### **CONNECT**
- Set up a tunnel to the user
- Brower  uses CONNECT to establish a secure TLS Channel

```
CONNECT api.example.com:443
```

### **TRACE**
- Echoback the recieved request (usually used for debugging)
- Is usually disabled

````python
#Find the total number of posts each user has written.
#We'll see the user format
main_url = "https://jsonplaceholder.typicode.com/users"

main_response = requests.get(main_url)
data = main_response.json()
ids = [i["id"] for i in data]

second_params = {"userId":ids}
second_url = "https://jsonplaceholder.typicode.com/posts"
second_response = requests.get(second_url, params = second_params)
print(second_response.url)
second_data = second_response.json()
````

````python
all_json = []
for i in ids:
    medium_dict = {}
    medium_dict["id"] = i
    posts = []
    for j in second_data:
        if j["userId"] == i:
            posts.append(j["title"])
    medium_dict["posts"] = posts
    medium_dict["num_posts"] = len(posts)
    for j in data:
        if j["id"] == i:
            medium_dict["name"] = j["name"]
            medium_dict["username"] = j["username"]
            medium_dict["email"] = j["email"]
    all_json.append(medium_dict)
````

````python
print(all_json)
````
