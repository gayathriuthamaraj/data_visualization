## **let's first create 15 stations and 20 trains**

```python
data_frame = pd.DataFrame(columns=["from", "to", "distance"])
```

```python
number_of_stations = 5
```

```python
data_frame.head()
```

```python
stations = []
for i in range(number_of_stations):
    stations.append(f"s_{i}")
```

```python
import numpy.random as nr
```

```python
#for the first 15 stations we'll do bidirectional
for i in stations:
    distance = nr.randint(1,30)
    to_station = nr.choice(stations)
    while to_station  == i:
        to_station = nr.choice(stations)
    data_frame.loc[len(data_frame)] = [i, to_station, distance]
    data_frame.loc[len(data_frame)] = [to_station, i, distance]
```

```python
data_frame.head()
```

```python
for i in range(number_of_stations):
    from_station = nr.choice(stations)
    to_station = nr.choice(stations)
    distance = nr.randint(1,30)
    while to_station  == from_station:
        to_station = nr.choice(stations)
    data_frame.loc[len(data_frame)] = [from_station, to_station, distance]
    data_frame.loc[len(data_frame)] = [to_station, from_station, distance]
```

```python
data_frame.head()
```

```python
data_frame.to_csv("stations.csv", index = False)
```

```python
import pandas as pd
import plotly.graph_objects as go

stations = list(set(data_frame['from']).union(data_frame['to']))
import numpy as np
angle = np.linspace(0, 2*np.pi, len(stations), endpoint=False)
pos = {station: (np.cos(a), np.sin(a)) for station, a in zip(stations, angle)}

edge_x = []
edge_y = []
for _, row in data_frame.iterrows():
    x0, y0 = pos[row['from']]
    x1, y1 = pos[row['to']]
    edge_x += [x0, x1, None]
    edge_y += [y0, y1, None]

edge_trace = go.Scatter(
    x=edge_x, y=edge_y,
    line=dict(width=1, color='#888'),
    mode='lines'
)

node_x = [pos[s][0] for s in stations]
node_y = [pos[s][1] for s in stations]

node_trace = go.Scatter(
    x=node_x, y=node_y,
    mode='markers+text',
    text=stations,
    textposition="top center",
    marker=dict(size=15, color='#00CC96')
)

fig = go.Figure(data=[edge_trace, node_trace])
fig.update_layout(
    title="Railway Network",
    showlegend=False,
    xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
    yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
)
fig.show()
```
