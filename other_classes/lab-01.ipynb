```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
plt.plot(x, y, marker='*', linestyle = '-', color='b')
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.savefig('sample_plot.png', dpi=300, bbox_inches='tight')
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
z = [1, 4, 9, 16, 25]
plt.plot(x, y, z, marker='1', linestyle = 'dotted', color='turquoise') #marker='d' for diamond and 'o for circle and 's' for square and 'x' for cross and 'v' for triangle and '>' for right triangle and '<' for left triangle and '^' for up triangle and 'p' for pentagon and 'h' for hexagon and 'H' for hexagon rotated and '|' for vertical line and '_' for horizontal line and '1' for tri down and '2' for tri up and '3' for tri left and '4' for tri right
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
```
