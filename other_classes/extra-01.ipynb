Data analysis usually refers to examining data to find patterns, trends or summaries -- often using basic statistics or visualization.

Data mining goes further; it uses advances algorithms (like machine learning, clustering, classification) to automatically discover hidden patterns, relationships or prediction in large datasets.

Data warehouse is a central place where a large amounts of data are stored


According to Comaford, instead of just considering the size of the dataset, one must include appropriate width, depth and volume. Effective data mining requires many attributes for the database records(width), a large number of records that are instances of database entities(depth), many entities determined by database design(volume).

Traditional statistics is assumption driven that a hypothesis formed first and validated by data however data mining is data driven that, hypothesis are formed automatically

data mining, the goal is to extract models which can be easily translated into logical rules or visual representations

Machine learning is the study of building systems that can learn from experience. It is designed to address problem domains that can be easily translated to address problem domains where good theoretical models does not exist

Furthermore, the machine learning paradigm "let the data suggest the model"

By monitoring  the performance of a subconscious skill, we may be able to construct a functional description of that skill in the form of symbolic rules

Neural networks are systems inspired by human brain. A basic example is provided by back propagation network which consists of input nodes, output nodes and intermediate nodes called hidden nodes. Initially, the nodes are adjusted with random weights. During the training, a gradient discent algorithm is used to adjust the weights so that the output nodes correctly classify data presented to the input nodes.

The trained network is viewed as a black box, since it does not give explanations to the results

Support vector machines can be geometrically interpreted

Tree based classifiers to break large dataset into smaller ones

Graphical models generalize markov models and hidden markov models

Rather than creating a single model, it is better to create a ensemble of models and choose using like a voting strategy (ensemble learning)

Linear algebra, recent work in parallel algorithms for solving linear system and algorithm for solving linear sparse systems with high dimensions important for data mining

Large scale optimization, sometimes the data mining algorithms can be expressed as large-scale and non-convex, optimization problems
