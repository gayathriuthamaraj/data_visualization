```python
plt.style.use('seaborn-v0_8-darkgrid')
plt.figure(figsize = (20,12))
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.subplot(2,1,1)
plt.plot(x,y1, 'g-', label = 'y1 = sin(x)')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Sin(x)', fontsize = 14)
plt.legend(loc = 'upper right')

plt.subplot(2,1,2)
plt.plot(x,y2, 'bo', label = 'y2 = cos(x)')
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.title('Cos(x)', fontsize = 14)
plt.legend(loc = 'upper right')

plt.show()
```

```python
x = np.arange(-50,50)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.sinh(x)
y5 = np.cosh(x)
y6 = np.tanh(x)
y7 = np.exp(x)
y8 = np.power(x,2)
y9 = np.power(x,3)

fig, axs = plt.subplots(nrows = 3, ncols = 3, figsize = (20,10))
axs[0,0].plot(x, y1, color = 'green', label = 'y1 = sin(x)')
axs[0,1].plot(x, y2, color = 'blue', label = 'y2 = cos(x)')
axs[0,2].plot(x, y3, color = 'pink', label = 'y3 = tan(x)')
axs[1,0].plot(x, y4, color = 'red', label = 'y4 = sinh(x)')
axs[1,1].plot(x, y5, color = 'orange', label = 'y5 = cosh(x)')
axs[1,2].plot(x, y6, color = 'yellow', label = 'y6 = tanh(x)')
axs[2,0].plot(x, y7, color = 'purple', label = 'y7 = exp(x)')
axs[2,1].plot(x, y8, color = 'magenta', label = 'y8 = power(x,2)')
axs[2,2].plot(x, y9, color = 'turquoise', label = 'y9 = power(x,3)')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

barWidth = 0.25

y1 = np.array([37, 45, 81, 7, 15])
y2 = np.array([15, 44, 51, 75, 34])
y3 = np.array([99, 78, 57, 100, 90])

pos1 = np.arange(0, 5)
pos2 = [x + barWidth for x in pos1]
pos3 = [x + barWidth for x in pos2]

plt.bar(pos1, y1, color="orange", width=barWidth, label="Open Tickets")
plt.bar(pos2, y2, color="red", width=barWidth, label="Waiting Tickets")
plt.bar(pos3, y3, color="yellow", width=barWidth, label="Closed Tickets")

plt.xlabel("Names")
plt.ylabel("Marks")
plt.xticks([i + barWidth for i in range(len(y1))], ["A", "B", "C", "D", "E"])

# ✅ Better formatted legend
plt.legend(
    title="Ticket Status",
    loc="upper left",   # "upper right"
    bbox_to_anchor=(1, 1),  # Places legend outside the plot
    fontsize=10,
    title_fontsize=12,
    fancybox=True,
    framealpha=0.7
)

plt.tight_layout()  # Helps layout adjust if legend is outside
plt.show()
```

```python
plt.style.use("classic")
# set width of bar
barWidth = 0.25

# set height of bar
y1= np.array([17,18,29,40])
y2 =np.array([20,21,22,23])
y3 =np.array([5,9,11,12])

# Set position of bar on X axis
pos1 = np.arange(len(y1))
pos2 = [x + barWidth for x in pos1]
pos3 = [x + barWidth for x in pos2]

# Make the plot
plt.bar(pos1, y1, color='#FBC02D', width=barWidth, label='Open')
plt.bar(pos2, y2, color='#F57F17', width=barWidth, label='Closed')
plt.bar(pos3, y3, color='#E65100', width=barWidth, label='Cancelled')

# Add xticks on the middle of the group bars
plt.xlabel('Assignee', fontweight='bold')
plt.ylabel('Number of Tickets', fontweight='bold')
plt.xticks([i + barWidth for i in range(len(y1))], ['Asif', 'Basit', 'Ravi', 'Minil'])

plt.legend(loc = "upper left")
plt.show()
```
