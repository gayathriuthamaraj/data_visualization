```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv'
iris = pd.read_csv(url)
iris.head()
```

```python
plt.plot(iris['sepal_width'][:30], marker='o', linestyle='--', color='blue')
plt.title('Sepal Length of First 30 Iris Samples')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
plt.plot(iris['petal_length'][:30], linestyle='-', color='blue', marker='x')
plt.title('Sepal Length (Saved Plot)')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.savefig('sepal_length_plot.png')
plt.show()
```

```python
plt.plot(iris['sepal_width'][:30], linestyle='dotted', color='pink', marker='d')
plt.title('Sepal Width (Saved Plot)')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Width (cm)')
plt.savefig('sepal_width_plot.png')
plt.show()
```

```python
plt.plot(iris['sepal_width'][:30], linestyle='-.', color='purple', marker='s')
plt.title('Sepal Width of First 30 Iris Samples')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Width (cm)')
plt.show()
```

```python
plt.plot(iris['sepal_width'][:30], linestyle='-', color='orange', marker='o')
plt.title('Sepal Width with Custom Axes')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Width (cm)')
plt.xlim(-5, 35)
plt.ylim(2, 5)
plt.xticks([0,2, 5, 10, 15, 20, 25, 30])
plt.yticks([2, 2.5, 3, 3.5, 4, 4.5, 5])
plt.show()
```

```python
plt.plot(iris['sepal_width'][:30], linestyle='--', color='teal', marker='^')
plt.title('Sepal Width ($cm$)', fontsize=16, fontweight='bold')
plt.xlabel('Sample Index', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.show()
```

```python
plt.plot(iris['petal_length'][:30], linestyle='-', color='green', marker='x')
plt.title('Petal Length of First 30 Iris Samples', fontsize=16, fontweight='bold')
plt.xlabel('Sample Index', fontsize=12)
plt.xlim(-5,35)
plt.ylim(0, 3)
plt.xticks([0, 5, 10, 15, 20, 25, 30])
plt.yticks([0,0.5, 1,1.5, 2,2.5, 3])
plt.ylabel('Petal Length (cm)', fontsize=12)
```

```python
x = iris['sepal_length'][:20]  # Selects the first 20 sepal length values from the iris dataset.
y = iris['sepal_width'][:20]   # Selects the first 20 sepal width values from the iris dataset.
#yerr = np.random.uniform(0.05, 0.2, size=20)  # Generates 20 random error values between 0.05 and 0.2 for error bars.
yerr = [0.9]
plt.errorbar(
    x, y, yerr=yerr,           # Plots y vs. x with vertical error bars defined by yerr.
    fmt='o',                   # Data points are shown as circles.
    color='green',             # Data points are green.
    ecolor='red',              # Error bars are red.
    capsize=3                  # Adds caps of size 3 to the error bars.
)
plt.title('Sepal Width with Error Bars')       # Sets the plot title.
plt.xlabel('Sepal Length (cm)')                # Labels the x-axis.
plt.ylabel('Sepal Width (cm)')                 # Labels the y-axis.
plt.show()                                     # Displays the plot.
```

```python
plt.scatter(iris['sepal_length'], iris['sepal_width'], color='blue', alpha=0.6)
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()
```

```python
plt.hist(iris['petal_length'], bins=20, color='purple', edgecolor='black', alpha=0.7)
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```
