```python
original_shuffled = df.sample(n = 1236, replace = True)
```

```python
mean_n = []
mean_s = []
```

```python
for i in range(150):
    temp = df.sample(n = 1236, replace = True)
    mean_n.append((temp.loc[temp["smoke"] == 0, "bwt"]).mean())
    mean_s.append((temp.loc[temp["smoke"] == 1, "bwt"]).mean())
```

```python
mean_n
```

```python
n_perm = 10000
perm_diffs = []

observed = df.loc[df['smoke']==0, 'bwt'].mean() - df.loc[df['smoke']==1, 'bwt'].mean()

for _ in range(n_perm):
    permuted_smoke = np.random.permutation(df['smoke'].values)
    mean_n = df.loc[permuted_smoke == 0, 'bwt'].mean()
    mean_s = df.loc[permuted_smoke == 1, 'bwt'].mean()
    perm_diffs.append(mean_n - mean_s)

p_value = np.mean(np.abs(perm_diffs) >= np.abs(observed))

print("Observed diff:", observed)
print("Permutation p-value (two-sided):", p_value)
```
