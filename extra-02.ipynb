```python
plt.figure(figsize = (12,5))
x = np.linspace(0,10,10000)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x,y1, 'g-', label = 'y1 = sin(x)')
plt.plot(x,y2, 'bo', label = 'y2 = cos(x)')
plt.xlabel('x')
plt.ylabel('Function Value')
plt.title("sin(x) and cos(x)", fontsize = 14)
plt.legend(loc='upper right',bbox_to_anchor=(1.2, 1.0))

plt.show()
```

```python
plt.style.use('seaborn-v0_8-darkgrid')
plt.figure(figsize = (20,12))
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.subplot(2,1,1)
plt.plot(x,y1, 'g-', label = 'y1 = sin(x)')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Sin(x)', fontsize = 14)
plt.legend(loc = 'upper right')

plt.subplot(2,1,2)
plt.plot(x,y2, 'bo', label = 'y2 = cos(x)')
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.title('Cos(x)', fontsize = 14)
plt.legend(loc = 'upper right')

plt.show()
```

```python
x = np.arange(-50,50)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.sinh(x)
y5 = np.cosh(x)
y6 = np.tanh(x)
y7 = np.exp(x)
y8 = np.power(x,2)
y9 = np.power(x,3)

fig, axs = plt.subplots(nrows = 3, ncols = 3, figsize = (20,10))
axs[0,0].plot(x, y1, color = 'green', label = 'y1 = sin(x)')
axs[0,1].plot(x, y2, color = 'blue', label = 'y2 = cos(x)')
axs[0,2].plot(x, y3, color = 'pink', label = 'y3 = tan(x)')
axs[1,0].plot(x, y4, color = 'red', label = 'y4 = sinh(x)')
axs[1,1].plot(x, y5, color = 'orange', label = 'y5 = cosh(x)')
axs[1,2].plot(x, y6, color = 'yellow', label = 'y6 = tanh(x)')
axs[2,0].plot(x, y7, color = 'purple', label = 'y7 = exp(x)')
axs[2,1].plot(x, y8, color = 'magenta', label = 'y8 = power(x,2)')
axs[2,2].plot(x, y9, color = 'turquoise', label = 'y9 = power(x,3)')
plt.show()
```

```python
x1 = [1,3,5,7]
x2 = [2,4,6,8]
y1 = [7,7,7,7]
y2 = [17,18,29,40]
plt.figure(figsize=(8,6))
ax = plt.axes()
ax.set_facecolor("blue")
plt.bar(x1,y1,label = "First",color = '#42B300') # First set of data
plt.bar(x2,y2,label = "Second",color = '#94E413', width=0.3) # Second set of data
plt.xlabel('$X$')
plt.ylabel('$Y$')
plt.title ('$Bar $ $ Chart$')
plt.legend()
plt.show()
```
